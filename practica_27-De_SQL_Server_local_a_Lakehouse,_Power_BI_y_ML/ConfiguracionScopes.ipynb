```python
from databricks.sdk import WorkspaceClient

# Crea un cliente autenticado con tu usuario del workspace
w = WorkspaceClient()

# NOMBRE DEL SCOPE QUE QUIERAS
scope_name = "mi-scope"

# Crear el scope (Databricks‑backed)
w.secrets.create_scope(scope=scope_name)

print("Scope creado:", scope_name)
```
