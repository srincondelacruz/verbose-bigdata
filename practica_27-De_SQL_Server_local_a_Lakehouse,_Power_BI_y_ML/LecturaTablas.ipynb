##Lectura de los archivos parquet en el storage account

```python
dbutils.secrets.listScopes()
```

```python
#Tabla clientes
df_clientes = spark.read.format("parquet").option(
    "fs.azure.account.key.stproyectoclase.dfs.core.windows.net",
    dbutils.secrets.get(scope="mi-scope", key="adls-key")
).load(
    "abfss://deltalake@stproyectoclase.dfs.core.windows.net/bronze/clientes/"
)

#Tabla productos
df_productos = spark.read.format("parquet").option(
    "fs.azure.account.key.stproyectoclase.dfs.core.windows.net",
    dbutils.secrets.get(scope="mi-scope", key="adls-key")
).load(
    "abfss://deltalake@stproyectoclase.dfs.core.windows.net/bronze/productos/"
)

#Tabla ventas
df_ventas = spark.read.format("parquet").option(
    "fs.azure.account.key.stproyectoclase.dfs.core.windows.net",
    dbutils.secrets.get(scope="mi-scope", key="adls-key")
).load(
    "abfss://deltalake@stproyectoclase.dfs.core.windows.net/bronze/ventas/"
)

```

```python
df_clientes.show(5)
df_ventas.show(5) 
df_clientes.show(5) 
```

##Creacion de tablas delta en el esquema creado

```python
# Guardar datos como tablas gestionadas (Lakehouse lógico)
#Clientes
df_clientes.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("azdb_proyecto_mario.lakeouse_proyecto.delta_clientes")

#Productos
df_productos.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("azdb_proyecto_mario.lakeouse_proyecto.delta_productos")

#Ventas
df_ventas.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("azdb_proyecto_mario.lakeouse_proyecto.delta_ventas")

```
