Ejercicio 1: Calculadora de Propinas
Calcula cuánta propina dejar en un restaurante.


```scala
def calcularPropina(cuentaTotal: Double, porcentaje: Int): Double = {
  cuentaTotal * porcentaje / 100
}

println(calcularPropina(50.0, 15))
```

Ejercicio 2: Años de Gato
Aquí convertimos la edad humana a años de gato multiplicando por 7.

```scala
def edadDeGato(edadHumana: Int): Int = {
  edadHumana * 7
}

println(edadDeGato(2))
```

Ejercicio 3: Repetidor de Palabras
Debemos repetir una palabra una cierta cantidad de veces. Truco de Scala: En este lenguaje, puedes multiplicar un texto (String) por un número para repetirlo automáticamente.

```scala
def repetir(palabra: String, veces: Int): String = {
  palabra * veces
}

println(repetir("Hola", 3))
```

Ejercicio 4: Mayor de Edad
Verificamos si la edad es 18 o mayor para saber si puede votar.

```scala
def puedeVotar(edad: Int): Boolean = {
  edad >= 18
}

println(puedeVotar(20))
```

Ejercicio 5: El Número del Medio
Tenemos que encontrar el valor intermedio entre tres números (a, b, c). La forma más fácil: Ponemos los números en una lista, la ordenamos de menor a mayor y tomamos el que queda en el centro (índice 1).

```scala
def numeroDelMedio(a: Int, b: Int, c: Int): Int = {
  List(a, b, c).sorted.apply(1)
}

println(numeroDelMedio(5, 2, 8))
```
