## 4. Operaciones de Agregación con Ejes (Reducción)

**Contexto:** Calcular métricas como el promedio de pérdidas o la desviación estándar por feature.

**Tareas:**
1. Crea una matriz `Data` de (5, 4) con valores aleatorios.
2. Calcula la **media** de toda la matriz en `mean_total`.
3. Calcula la **suma** de cada columna (a lo largo del eje 0) en `sum_cols`.
4. Encuentra el **valor máximo** de cada fila (a lo largo del eje 1) en `max_rows`.


```python
np.random.seed(42)
Data = np.random.rand(5, 4)

print("--- Matriz Data (5x4) ---")
print(Data)

# 2. Media total
mean_total = Data.mean()
print(f"\nMedia total de la matriz: {mean_total:.4f}")

# 3. Suma por columnas (axis=0)
sum_cols = Data.sum(axis=0)
print("\n--- Suma de columnas (axis=0) ---")
print(f"Valores: {sum_cols}")
print(f"Forma resultante: {sum_cols.shape}") # Debería ser (4,)

# 4. Máximo por filas (axis=1)
max_rows = Data.max(axis=1)
print("\n--- Máximo de cada fila (axis=1) ---")
print(f"Valores: {max_rows}")
print(f"Forma resultante: {max_rows.shape}") # Debería ser (5,)
```

## 5. Filtrado con Máscaras Booleanas (Limpieza de Datos)

**Contexto:** Necesitas eliminar outliers o filtrar valores específicos, una técnica muy usada en preprocesamiento.

**Tareas:**
1. Genera un vector `arr` con 20 números aleatorios entre -10 y 10.
2. Crea una "máscara" booleana `mask` que sea `True` donde los valores de `arr` sean negativos.
3. Usa la máscara para reemplazar todos los valores negativos de `arr` por 0 (implementando una ReLU manual en su lugar).
4. Cuenta cuántos valores del array original eran mayores a 5.


```python
np.random.seed(42)
arr = np.random.randint(-10, 11, 20)

# Guardamos copia para el punto 4
arr_original = arr.copy()

print("1. Array original:\n", arr)

# 2. Crear máscara
mask = arr < 0
print("\n2. Máscara (True donde es negativo):\n", mask)

# 3. Aplicar máscara (ReLU)
arr[mask] = 0
print("\n3. Array modificado (Negativos -> 0):\n", arr)

# 4. Contar mayores a 5 en el original
# Vemos el array original: [ 6, 9, 4, -3, -4, ... ]
conteo = np.sum(arr_original > 5)
print(f"\n4. Cantidad de valores > 5 en el original: {conteo}")
```

## 6. Multiplicación Matricial vs Element-wise

**Contexto:** Diferenciar entre el producto de matrices (operación lineal) y el producto elemento a elemento (Hadamard).

**Tareas:**
1. Crea dos matrices `X` e `Y`, ambas de 3x3 (usa `np.arange` para tener valores predecibles).
2. Calcula el **producto punto** (multiplicación matricial) en `dot_product` (usa `@`).
3. Calcula el **producto elemento a elemento** en `element_wise_product` (usa `*`).


```python
# 1. Crear matrices
X = np.arange(9).reshape(3, 3)
Y = np.arange(9).reshape(3, 3)

print("Matriz X (y Matriz Y son iguales):\n", X)

# 2. Producto Punto (@)
# Ejemplo cálculo celda [0,0]: (0*0) + (1*3) + (2*6) = 0 + 3 + 12 = 15
dot_product = X @ Y
print("\n--- Producto Punto (X @ Y) ---")
print(dot_product)

# 3. Producto Elemento a Elemento (*)
# Ejemplo cálculo celda [0,0]: 0 * 0 = 0
# Ejemplo cálculo celda [2,2]: 8 * 8 = 64
element_wise_product = X * Y
print("\n--- Producto Elemento a Elemento (X * Y) ---")
print(element_wise_product)
```

## 7. Manipulación de Dimensiones (Flatten y Reshape para CNNs/RNNs)

**Contexto:** Preparar imágenes para una red densa o adaptar tensores para capas convolucionales.

**Tareas:**
1. Crea un tensor aleatorio `imagenes` con forma `(10, 28, 28)` (Batch, Alto, Ancho).
2. Aplana las dimensiones espaciales para obtener una forma `(10, 784)` en `flattened`.
3. Añade una dimensión extra al final para representar el canal (blanco y negro), quedando `(10, 28, 28, 1)` en `reshaped_channel`.


```python
# 1. Crear tensor
np.random.seed(42)
imagenes = np.random.rand(10, 28, 28)

print(f"1. Forma original: {imagenes.shape}")
# Salida: (10, 28, 28)

# 2. Aplanar (Flatten)
# Esto simula preparar los datos para una red neuronal densa (MLP)
flattened = imagenes.reshape(10, 784)

print(f"2. Forma aplanada: {flattened.shape}")
# Salida: (10, 784)

# 3. Añadir canal
# Esto simula preparar los datos para una red convolucional (CNN) en Keras/TensorFlow
reshaped_channel = imagenes.reshape(10, 28, 28, 1)

print(f"3. Forma con canal: {reshaped_channel.shape}")
# Salida: (10, 28, 28, 1)
```

## 8. Vectorización de una Función (Evitar Bucles)

**Contexto:** Implementar lógica condicional de manera eficiente sobre arrays grandes.

**Tareas:**
1. Define un array `x` con 10 números aleatorios (algunos positivos y algunos negativos).
2. Usa la función `np.where` para crear un array `output` donde: 
   - Si el valor en `x` es mayor a 0, se mantiene ese valor. 
   - Si el valor en `x` es menor o igual a 0, se reemplaza por el valor absoluto de ese número.


```python
np.random.seed(42)

# 1. Definir array x con 10 números aleatorios (positivos y negativos)
# Usamos randn para distribución normal (centrada en 0), así garantizamos +/-
x = np.random.randn(10)

print("Array original x:\n", x)

# 2. Usar np.where para crear output
# Condición: x > 0
# Si es True: mantén x
# Si es False: usa np.abs(x) (o -x)
output = np.where(x > 0, x, np.abs(x))

print("\nArray output (todos positivos):\n", output)
```
