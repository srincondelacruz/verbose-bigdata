```python
import pandas as pd
```

Choosing data in a Series
Let’s create a simple series and store it in a variable named “series_1”.

```python
series_1 = pd.Series([2,3,4,5,6])
series_1
```

The output of the series shows index 01,2,3,4 on the left column. We can use these index values to call a data value in the series.

Series - Selecting data based on the index

If we want to get the value stored at Series index position 3, we would type the following code, and Pandas will return the value 5 stored at index value 3.

```python
series_1[3]
```

Series — Re-indexing a series to make things easier

Pandas also allows you to create an index with string (text) values. In some cases, it might be easier to call an index position by its name.

So let’s start a series where our index will be candy names, and the data will reflect candy quantities.

```python
sale = pd.Series([500,85000,40000,9600,100000], 
                 index=['Cookie', 'Choco','Toothpaste','Soap','Toffee'
                        ])
sale
```

Now let’s apply the same approach, but this time to the new index we have created.

```python
sale.Toffee
```

Choosing data in Data frame
As discussed before, a data frame is a two-dimensional data structure with rows and columns. The Rows and columns in a data frame are indexed separately and can be used to select the data we need for our analysis.

Let’s create a simple data frame with three columns state, year and pop (population).

```python
data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada', 'Nevada'],'year': [2000, 2001, 2002, 2001, 2002, 2003],'pop': [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}
data = pd.DataFrame(data)
data
```

Selecting columns in a data frame

In our example, we will select the column ‘state’ from the data frame, and the result will show us all the data stored in that column. There are two ways to select the column.

Dot notation method

```python
data.state
```

```python
data.state
```

Choosing either method is a personal choice, but there are some caveats that one should keep in mind.

Dot notation has some limitations:

It does not work if the column name has spaces.
The column name is a python or pandas keyword.
The column name is an integer.
You can not add a new column name with do notation.
Now let’s see a practical example of why dot notation is limited in some instances. We aim to select a pop (population) column with dot notation.

```python
data.pop
```

This output is not what we expected because ‘pop’ is a python keyword/function.
In this scenario, using the Index Operator method would be appropriate. Dot notation method cannot process the word pop as it is a python keyword and displays the column selection accurately.

```python
data['pop']
```

In summary, it is a matter of choice; some consider Dot notation to be convenient as it takes only one dot compared to [“ “] four extra characters.

We also can select multiple columns by defining a list of column names.

```python
data[['pop', 'state']]
```

Keep in mind that the order defined in the list is shown in the output. And it might be different from the order in the original dataframe.

Selecting data frame rows and columns using .iloc and loc

Pandas loc and iloc methods are two top choices when selecting data using rows and columns.

loc is label-based. You can select rows with index, but only labels are used to choose columns.
iloc is an integer location-based. We can select only rows and columns with their index positions.
The basic usage of loc and iloc is df.iloc[rows , columns] and df.loc[rows, columns]. The difference is passing the index position or labels.

We will use a real-world data set called Stack Overflow Annual Developer Survey for this section. It can be found here. (Choose the year 2020 data set)

Let’s load the entire data set with all 61 columns.

```python
df = pd.read_csv('../survey_results_public.csv')
df
```

Or, for simplicity, we can also load the data set with just the columns we need. In this case, we only want the Age, ConvertedComp, Country, LanguageWorkedWith and YearsCode columns. The total number of columns are in the data set is 61.

```python
df = pd.read_csv('../survey_results_public_2020.csv', usecols=['Age','ConvertedComp', 'Country', 'LanguageWorkedWith', 'YearsCode'])
df
```

iloc

Let’s get the value stored at row number 1 and column number 2 using iloc.

```python
df.iloc[1, 2]
```

Selecting multiple rows can be difficult; a shorter way is to choose a range of index positions. Here we choose a slice of rows for 2 to 9 using the range (:).

Remember that using the range of rows will have the start index as inclusive, and the end index is exclusive. So it will give values from rows 2 to 8 of column number 2.

```python
df.iloc[2:9, 2]
```

Let’s choose data in all five columns for rows 1 to 49999.

```python
df_subset = df.iloc[1:50000]
df_subset
```

Here we can observe that column name is not mandatory, and we have to write index position 50000 to select rows up to number 49999.

loc

As discussed loc function can use either index position or label for rows. So let’s select data in column LanguageWorkedWith from 500 to 550.

But wait a minute, why did we not write row 551 to get row 550 in the loc function? Here it is important to remember the loc function works on labels, even integer-based locations treated as labels. That’s why the output we will get from 550 to 550, unlike the iloc function from 550 to 549.

```python
df.loc[500:550, 'LanguageWorkedWith']
```

Let’s expand our selection to multiple rows and columns with the loc function.

```python
df.loc[[1,2,3,4,5], ['Age', 'Country', 'ConvertedComp']]
```

Bonus Section
Bonus Section So far, we have worked with rows with their index position numbers. We may want to use a column as the index in some scenarios. Let’s convert the Country column to index. Remember that the country column data will now be in the index, and there will be no more Country column in the data frame.

```python
df.set_index('Country', inplace=True)
df
```

Now let’s select the data belonging to Morocco and the United States in columns Age to LanguageWorkedWith. Note we are also able to use the range with columns.

```python
df.loc[['Morocco', 'United States'], 'Age':'LanguageWorkedWith']
```
