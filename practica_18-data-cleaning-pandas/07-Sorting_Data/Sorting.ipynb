Sort based on a single column
The sort_values() method is used to arrange the data along their axis (columns or rows) in the Pandas data frame.

Sort_values() method parameters:

by: It takes a single column or list of columns to sort by.
ascending: By default, set to True. If we want to sort in — descending, we set ascending as equal to False.
kind: This parameter takes a sorting algorithm. By default, ‘quicksort’ is set. However, we can change to ‘mergesort’ and ‘heapsort’. Quicksort is the fastest. Merge sort and Heap sort are both efficient for larger data sets. Merge sort is a little quicker but requires mode memory.
inplace: By default, set to False. When set to True, it applies changes in our original data frame.
Let’s sort our data by country column.

```python
df.sort_values(by = 'Country')
```

Notice the Country column has been sorted. Now lets do a Country column descending sort.

```python
df.sort_values(by = 'Country', ascending=False)
```

Sort based on multiple columns
So far, we have sorted our data based on a single column, but Pandas allows us to sort based on multiple columns.

It can be done with the sort_values() method, but we pass the list of columns.

Here we sort our data bytwo columns Country and ConvertedComp. First, it sort by Country column with ascending order, and then the ConvertedComp column.

```python
df.sort_values(by = ['Country', 'ConvertedComp'])
```

Notice Country columns is sort in the ascending order and the ConvertedComp column is also in ascending order with respect to Country column.

By default, both columns are sorted in ascending order, but we can change the order of both by passing a list of boolean operators in ascending parameters.

Lets sort the Country column is sorted by ascending order, and ConvertedComp is by descending order.

```python
df.sort_values(by = ['Country', 'ConvertedComp'], ascending = [True, False])
```

Sort based on index
We can sort indexes (rows or columns) with the sort_index() method. This method is used to sort data by index labels. By default, it sorts by rows, but when the axis parameter is set to 1 (axis = 1), it sort by columns.

```python
df.sort_index()
```

To sort in descending order

```python
df.sort_index(ascending=False)
```

We can also sort column names in order.

```python
df.sort_index(axis='columns')
```

More Sorting methods
nlargest(): The nlargest() method returns the first n largest values in descending order. Where n is the number we give.

nsmallest(): The nsmallestt() method returns the first n smallest values in ascending order. Where n is the number we give..

These methods can be used with either series or data frame. Let’s use these methods practically.

Lets see the first 10 largest values of ConvertedComp column.

```python
df['ConvertedComp'].nlargest(10)
```

Now if we want to see the rows that correspond to the first 10 largest values of Age column.

```python
df.nlargest(10, 'Age')
```

Similarly let see the 10 smallest values of Age column

```python
df['Age'].nsmallest(10)
```
