What are we going to learn today?
In this article, we will learn to update and substitute the values in Pandas data frames.

Why is it important to learn?
Often real-world data sets are not conducive to analysis, for example — Incorrect values, syntax, and mistakenly inserted records. That’s why updating rows and columns are necessary to learn to make our data analysis easy.

We will see an example of students marks and update a column with Pass or Fail when the data values meet passing marks threshold criteria.

How are we going to achieve today’s learning goal?
Plan for today is:

Updating column names
Updating rows
Updating rows based on conditions

```python
import pandas as pd
```

We will create a data frame with the help of a dictionary.

```python
Student = {
"Id": ['cs-01', 'mth-02', 'eng-03', 'phy-04', 'mth-05', 'cs-06'],
"Name": ["Fatima Khan", 'Noureen Ali', 'Salman Aslam', 'Sana Khan', 'Adam Jack', 'Ali Khan'],
"subject": ['computer science', 'math', 'english', 'physics', 'statistics', 'programming'],
"marks": [45, 90, 60, 37, 49, 50]
}
data = pd.DataFrame(Student)
data
```

Updating column names
The rename method is used to alter the names of the columns in the data set. It takes the columns parameter, and we will assign a dictionary to this parameter, where the dictionary’s key is the old column name, and the value will be the new column name.

Let’s change the Name column into Full Name as it is more appropriate. And to make the change permanent, we will set the inplace parameter to True. By default, it is set to False.

```python
data.rename(columns = {'Name': 'Full Name'})
```

Now, we will change all the column names into the lower case, which will make it easy to select columns. We will use the columns.lower() method with string class (str class). Without using the string class we will get an error.

```python
data.columns = data.columns.str.lower()
data
```

Updating Rows
Now let’s update a row. First, let’s select row number four.

```python
data.loc[4]
```

Now we will create a list with the new row values and assign it to row 4.

```python
data.loc[4] = ['cs-05', 'Ali', 'Database', 47]
data.loc[4]
```

We can see the output that the row has been updated. Very easy.

What if we wanted to update a few values, not a complete row. We can achieve it by selecting only the values we want to edit. To update the name and subject in row 0, we will select it first.

```python
data.loc[0, ['name', 'subject']]
```

We assign two new values by selecting row 0 with the name and subject columns.

```python
data.loc[0, ['name', 'subject']] = ['Farhan', 'Python']
data.loc[0]
```

Updating Rows based on conditions
Updating rows based on certain conditions is a widespread use case. We will update the marks column with a Fail string when the value is below 50.

First, let’s create a condition and assign it to a variable

```python
filt = data['marks'] < 50
filt
```

The output returns a boolean value. It returns True where the condition is matched and False where unmatched. To get the desired output, we will go to our trusty loc method and use the condition filter (filt) we created to update the marks column.

```python
data.loc[filt, 'marks'] = 'Fail'
data
```

Awesome! Now, we can easily see who failed the test.
