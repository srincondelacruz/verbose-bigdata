v)Replace with mode,mean and median- the best solution is to replace with median value for that column. Median is a good choice because it is least likely to influence the distribution of the dataset overall. If the data is symmetric, then the mean and median will be the same value. If the dataset is not symmetric, then the mean is more likely to be skewed by outlier values, so the median is a better choice.

Mean
-Age is a string and when you run the mean function directly you will get a Type error therefore we need to convert Age from string to numeric same way we converted sale_date to date earlier

df[‘Age’]= pd.to_numeric(df[‘Age’])
df.dtypes

```python
df.Age =pd.to_numeric(df.Age)
df.dtypes
```

We can see that age has been converted to a float we can now replace the NaN with mean

```python
df
```

```python
x = df.Age.mean()
df.Age = df.Age.fillna(x)
df
```

You can see that now NaN on Age column has been filled with mean

Mode and Median
-The formula will be the same only that on the value of x you will replace with mode or median i.e

Mode

Median

```python
x =df.Age.mode()
df.Age = df.Age.fillna(x)
df
```
