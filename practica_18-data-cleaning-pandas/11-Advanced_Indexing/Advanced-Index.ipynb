What are we going to learn today?
In this article, we will learn multi Indexing in Pandas. So far, we have seen data sets with one-level index, but in multi-level or advanced indexing, we have a hierarchy of indexes. We will learn different ways of creating and manipulating multi-index in Pandas.

Why is it important to learn?
Multi-level indexing provides granularity. It provides an easier, more accurate and faster way to filter and process information. For example, a football player can be referenced with a hierarchical index of (Country, State, Ranking).

Multi indexing is an excellent way to analyze and manipulate sophisticated data, especially working with higher dimensional data.

```python
import pandas as pd
import numpy as np
```

```python
data = pd.Series( [56,78,99,56,88,59,100,57,91], index=[['Team1', 'Team1', 'Team1', 'Team2', 'Team2', 'Team2','Team3', 'Team3', 'Team3'],['player_1', 'player_2', 'player_3','player_1', 'player_2', 'player_3','player_1', 'player_2','player_3']],name="Player's Score")
data
```

Let’s review the output, and we can see that we have two indexes. The first index is called level 0 index and the 2nd index is level 1 index.

Now let’s view just the multi-index of the series.

```python
data.index
```

Let’s select a range by selecting level 0 index.

```python
data['Team1' : 'Team2']
```

Multi-index behaves differently with the loc and iloc function. Level 0 is treated as a row and Level 1 is treated as a column.

For example, if we want to select all level 0 indexes but only index number 2 value of level 1 index. We can writethe code like this.

```python
data.loc[:, 'player_1']
```

Here let’s select a specific value by supplying the exact indexs.

```python
data.loc['Team1', 'player_2']
```

Data frame with Multi-Index
Now, we will see a multi-index in data frame. For that, we will create a small data frame with a dictionary.

```python
df = {'name' :['Jack', 'Fatima', 'Salman', 'Sahil', 'Noureen', 'Hamza'],'rank' :[1, 2, 1, 2, 1, 2],'city' :['Khi', 'Khi', 'Isb', 'Isb', 'Lhr', 'Lhr'],'CS_rating' : [21, 56, 34, 45, 24, 34],'Maths_rating' : [76, 45, 56, 34, 234, 45]}
df = pd.DataFrame(df)
df
```

Now lets set the city and rank columns as multi index using the set_index method. Moreover, set the inplace parameter is set to True so our original data frame will be changed permanently.

```python
df.set_index(['city', 'rank'], inplace=True)
df
```

And now we can see that our data has multi-index with city and rank as indexes.

```python
df.index
```

Let’s suppose we want the record stored at multi index Khi, whose rank is 1 using the loc method.

```python
df.loc['Khi', 1]
```

Reshaping a data frame
When we have multi-index, we can use stack and unstack methods which returns a reshaped data frame.

unstack(): It returns a data frame similar to a pivot table in excel. It returns a new level of columns whose inner-most level consists of the pivoted index labels.

stack(): It returns a reshaped multi index Series.

Let’s review unstack() first.

```python
df.unstack()
```

Let’s review the output, we have new column rank and row labels city. We can interpret the output that Salman’s rank is 1 and Sahil’s rank is 2. They are both from Isb, and Salman’s score_1 is 34, and Sahil’s score_1 is 45 same as score_2.

Now let’s review the stack() method.

```python
df.stack()
```

As discussed, stack()it returns a multi index series.

Sorting in Multi-Index
Index sorting is important in real life. We will use the sort_index() method with level parameter. In our case city is level 0 index and rank is level 1 index.

```python
df.sort_index(level=0)
```

The city is on level 0, and the rank is on level 1.

```python
df.sort_index(level=1)
```

The output shows that the city index is sorted in ascending order when the level is set to 0, and rank is sorted in ascending order when the level is set to 1.
