What are we going to learn today?
In this article, we will learn how to remove /edit missing and incorrect values in a data set. This process is considered part of data cleaning.

In Pandas, missing values are called null values (NumPy Not a Number-NaN). Incorrect values can be anything that must be removed or edited for practical data analysis.

NumPy is an advanced mathematical functions library, and we will use it for creating null values in pandas np.nan, which will convert into NaN.

Why is it important to learn?
The majority of the time, data scientists are busy with data cleaning. Because, in the real world, data is not what we want it to be. Real-world data is noisy, which affects the accuracy of our analysis and machine learning models. We will learn more about machine learning later, but here is a quick overview of why data cleaning helps with machine learning.

Machine learning models perform very poorly on noisy and uncleaned data. The poor model means it leads to underfitting and overfitting.

Machine learning models are algorithms that train over a set of data to recognize certain types of patterns or make predictions.
Underfitting means the machine learning model neither performs well on training nor test data.
Overfitting is when a machine learning model performs well on training data but poor on testing data.

```python
import pandas as pd
import numpy as np
```

We will create a custom dataset with missing values — NaN and None. The np.nan (NaN) and None values are null values in pandas.

Apart from np.nan and None, we will add some incorrect data with string values like (‘NaN’, ‘None’, ‘?’). It will seem like null values in the data frame, but these are actually strings.

Let’s start by creating a dictionary as the basis for our data frame.

```python
student = {
'first': ['Fatima', 'Carine', 'Alla', 'Salman', np.nan, None, 'NA'],
'last': ['Khan', 'Aries', 'Ben', 'Malik', np.nan, np.nan, 'NaN'],
'email': ['fatimakhan@gmail.com', 'caries@email.com', 'allaben@email.com', None, np.nan, 'saqib@gmail.com', 'NA'],
'age': [33, 55, 63, 36, np.nan, None, '?']
}
student
```

Converting the dictionary into a data frame and display it.

```python
data = pd.DataFrame(student)
data 
```

Handling null values
First, we will check if there are any null values in the data set with the isnull() method. It returns a boolean operator True where it finds a null value; otherwise, it returns False.

```python
data.isnull()
```

If we want to count the number of null values, we will sum it with the sum() method.

```python
data.isnull().sum()
```

We can see several null values in the data frame. To remove null values, we will be using dropna() method. dropna(): It removes missing values from the data set. Important parameters in this method are:

axis: It removes null values from either rows or columns. axis = 0 or index means rows and axis = 1 or columns means columns.

how: Using this, we describe how to remove null values. If how is set to any, it means drop that row or column where at least one null value. If how is set to all, it drops the row and column where all values are null. By default, how is set to rows.

inplace: It takes boolean operators. If it is True, we want to make a permanent change in the original data frame. If it is False, it only shows the result of the dropna() method without making any changes to the data set.

```python
data.dropna(how='any')
```

Let’s change the axis, and Now we will remove by columns.

```python
data.dropna(how='any', axis=1)
```

Now let’s try a combination of all and axis set to 1

```python
data.dropna(how='all', axis=1) 
```

The result shows the full data frame as no column has all null values.

Now let’s try all and the axis will be set to rows by default.

```python
data.dropna(how='any')
```

Let’s review the output. Interestingly row 6 is still present while row 4 is gone. This is because row 6 has fake null values :). These are strings and not real null values.

On to the next section to learn more about custom null values aka fake null values.

Dealing with custom missing values
Here we are using the NaN values and set the inplace parameter True — row 6 has custom missing values, and we will change them into real missing values.

```python
data.replace({'NA':np.nan, 'NaN':np.nan, '?':np.nan}, inplace=True)
```

Let’s execute the dropna() method now to drop all rows with null values including row 6.

```python
data.dropna()
```

Let’s execute the dropna() method now to drop all rows with null values including row 6.

Filling in missing data
Sometimes removing missing values is not effective as we might lose important information. It can affect our machine learning model’s performance as well. So instead of removing, we fill missing numerical values with central tendency values (mean, median and mode). To fill missing values, we used the fillna() method.

The fillna() method replaces the missing values with a specified value. It also takes a few parameters:

inplace: Changes in the original data frame.

axis: Axis along which to fill missing values.

For example, we will add a MISSING string in the data frame without mentioning an axis. It will fill all the missing values with MISSING.

```python
data.fillna('MISSING')
```

Now let’s fill the MISSING values in the age column with the mean of the columns instead of removing the MISSING values.

```python
data['age'].fillna(data['age'].mean(), inplace=True)
data
```

Now, let’s see the output, and we can see that the age column MISSING values are filled by its mean.
