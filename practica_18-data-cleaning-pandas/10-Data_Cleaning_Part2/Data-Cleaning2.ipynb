Interpolation Methods
In statistics, interpolation is a method in which known values are used to estimate unknown values. Pandas, ffill and bfill methods for interpolation. ffill and bfill are sub-methods of the fillna() method, used to fill missing values.

ffill: ffill means forward to fill. It is used to forward fill the missing values present in the data set.

bfill: bfill means backward fill. It is used to backwards fill the missing values present in the data set.

Let’s create a custom data set with missing values. We will use the random function from the NumPy library and fill some NaN values using the iloc method.

```python
data = pd.DataFrame(np.random.randn(7,3))
data
```

Let’s sprinkle some NaN in the data set to simulate missing values.

```python
data.iloc[:3, 1] = np.nan
data.iloc[:2, 2] = np.nan
data.iloc[4:6, 2] = np.nan
data
```

We will use the ffill method first to fill the missing values in the data set.

```python
data.fillna(method='ffill')
```

The output shows that rows 4 and 5 of column 2 have been filled with the value of row number 3 and column number 2. You might be wondering why we still have missing values in the data set. Go back to the definition of the ffil method, and you will see that it only forwards fills the values.

The backward fill method is as easy as the forward fill. Let’s see the data frame real quick.

```python
data
```

Data frame is back to its original state.

Apply bfill.

```python
data.fillna(method='bfill')
```

The output shows the missing values have been backwards filled.

We can also use a limit parameter with ffill and bfill methods. It controls the number of values the method would fill.

Let’s set the limit to 1 and use the ffill method.

```python
data.fillna(method='ffill', limit=1)
```

The output shows that only row number 4 of column 2 is filled with the value.

Removing Duplicates
One of the main problems we get from duplicated data is data leakage. Removing duplicates isessential for high performing machine learning models and better data analysis.

We will use the duplicated() method to remove the duplicated records.

First, we will check our StackOverflow survey data set. We will create a custom dataset to understand the duplicated() method if we do not find any.

Let’s see how many duplicates we have in our data set.

```python
df.duplicated().sum()
```

Since our existing data set has no duplicated we will create one with duplicates.

Creating a data frame from the dictionary.

```python
data1 = pd.DataFrame({'col1': ['one', 'two'] * 4 + ['two'], 'col2': [1, 1, 2, 3, 3, 4, 4,5,5]})
data1
```

Let’s check the duplicate records. At first, it returns boolean values.

```python
data1.duplicated()
```

We can get a count of duplicate values with the sum() method.

```python
data1.duplicated().sum()
```

We have found 1 duplicate record. To remove duplicated records, drop_duplicates() methods are used. It also takes inplace parameters if we want to change in the original dataframe.

```python
data1.drop_duplicates()
```

We can see in the output that we got 7 rows instead of 8.
