Column condition filter
Let’s extract data based on the condition of when the salary is more than 40000 USD dollars in the ConvertedComp column.

```python
df['ConvertedComp'] > 40000
```

The output gives boolean values instead of records. It means where it finds condition is matched, it returns True, and if not, it returns False.

In order to get rows with data that meets our condition, we have to apply the condition in a data frame, and this will give us our expected result.

```python
df[df['ConvertedComp'] > 40000]
```

Column and row condition filter
We can also pass conditions with the loc and iloc methods, which is a more convenient way to filter out the data because we can get only those rows & columns we are interested.

This time we create the same condition but assign it in a variable (high_salary_filter) that increases the readability, and our code looks better.

```python
high_salary_filter = df['ConvertedComp'] > 40000
```

Now we will pass our condition to the loc method and select only three columns — Country , ConvertedComp, LanguageWorkedWith

```python
df.loc[high_salary_filter, ['Country', 'ConvertedComp', 'LanguageWorkedWith']]
```

The output gives a concise result that meets both row and column requirements.

Inverted condition filter
~ : Tilde (~) is a Python negation operator which reverses the condition. For example, ~ON becomes OFF.

Now, if we want to see the developer whose salary of less than 40000, instead of writing the condition again, we just put a ~ operator on a high_salary_filt (~ high_salary_filt). Isn’t that awesome….

```python
df.loc[~ high_salary_filter, ['Country', 'ConvertedComp', 'LanguageWorkedWith']].head(20)
```

NaN means Null Values. It means respondents did not answer this question or they are students with no salary.

Important to remember here is that NaN is considered a value less than the stated value. Apart from this, we only got those developers whose salary is less than 40000.

Complex filter using multiple conditions
This section will use Logical (AND, OR) Operators. Before jumping to the complex filtering, I want to introduce you to logical operators AND and OR operators.

AND: It returns TRUE only if all conditions are true.
OR: It returns TRUE only if either one operand is true.
In Pandas, & is equal to AND and | is equivalent to OR
In our example, the filter is defined as follows:

Salary of greater than 50000 in ConvertedComp

AND

The Country column should contain value to the United States

AND

LanguageWorkedWith column string values contain ‘Python.’ — (we will use str. contains() method for this condition. This is a new method we are learning here. It is used to test if a pattern or sub string is contained within a string value.)

All these conditions will be applied to the loc method as a filter, and the rows that meet the requirements should return columns ‘Country’, ‘ConvertedComp’, ‘LanguageWorkedWith’.

```python
filt = ((df['ConvertedComp'] > 50000) & (df['Country'] == 'United States')) | (df['LanguageWorkedWit'].str.contains('Python'))
df.loc[filt, ['Country', 'ConvertedComp', 'LanguageWorkedWith']].head(50)
```

```python
filt = ((df['ConvertedComp'] > 50000) & (df['Country'] == 'United States')) | (df['LanguageWorkedWith'].str.contains('Python', na=False))

df.loc[filt, ['Country', 'ConvertedComp', 'LanguageWorkedWith']].head(50)
```

Let’s do another example; if we want to analyze the rows belonging to three Asian countries, we can create a filter using the OR (|) condition that filters out the rows belonging to three Asian Countries (India,China,Indonesia).

```python
Asia_country = (df['Country'] == 'India') | (df['Country'] == 'China') | (df['Country'] == 'Indonesia')
```

Now we are storing this data into another variable and printing the first five rows of it.

```python
Asian_df = df.loc[Asia_country]
Asian_df.head()
```

The output contains rows with unique values of the country column (China, India, Indonesia). Let’s use the unique method to verify our result.

```python
df.loc[Asia_country, 'Country'].unique()
```

Bonus Section
So far, we saw multiple conditions using the OR (|) operator, which is very helpful but let suppose if we want to filter out 10 Asian Countries, it will be a very long syntax. At the same time, Python gives a great isin() operator, which checks whether values are contained in Series.

In other words, it shortens the OR condition.

```python
Asia_country = df['Country'].isin(values = ['China', 'India', 'Indonesia']) # Passing a list to values parameter
```

now lets review the unique values

```python
df.loc[Asia_country, 'Country'].unique()
```

Interesting, you can see this is a fancy and less time-consuming way to write multiple OR conditions.
