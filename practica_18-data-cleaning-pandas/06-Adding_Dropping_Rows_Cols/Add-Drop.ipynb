Adding columns with Indexing Operator
This section will add a Carbohydrates Percentage column to the data set. We will achieve this result by applying the percentage formula to each value in the Carbohydrates and Calories column. The result will be saved in the new column.

Formula

Carbohydrates Percentage = (Carbohydrates/Calories) x 100

It is the first method to create/add a new column, write a new column name in indexing operator [ ] and assign a new value/condition.

```python
df['Carbohydrates Percentage '] = df['Carbohydrates'] / df['Calories'] * 100
```

Let us display the columns relevant to our goal and see the new column Carbohydrates Percentage with the percentage of carbohydrates in each item.

```python
# Fíjate en el espacio extra al final de Percentage
df.loc[ : , ['Category', 'Item', 'Calories', 'Carbohydrates', 'Carbohydrates Percentage ']].head()
```

A useful insight for those who need to manage their carbohydrates intake.. :)

Adding columns with Insert method
This section will add a Protein Percentage column to the data set using the insert method. We will achieve this result by applying the percentage formula to each value in Protein and Calories column. The result will be saved in the new column.

Formula

Protein Percentage = (Protein/Calories) x 100

Insert method is used to insert or create a column at a specified location. It takes four parameters, location where you want to insert a new column, new column name, value and allow_duplicates whether you want to allow duplicates column name or not.

```python
df.insert(4, 'Protein Percentage', df['Protein'] / df['Calories'] * 100, False)
```

Lets view a few columns.

```python
df.head()
```

Now we can see that column at index number four is Protein Percentage.

Removing/Dropping Columns
Let us remove/drop the column Protein in percentage, and if we want to drop more than one column, we will pass a list of those columns.

For dropping columns in Pandas data frame, we will use the drop() method.
The drop() method is used to drop/remove specified columns. In order to change in the original data frame, we will use the inplace=True parameter.

```python

df.drop(columns='Protein Percentage', inplace=True)
df.head()
```

Let’s display the five rows of the data frame, and you will see there is no column Protein percentage at index number 4.

Adding rows with Concat method
So far, we have seen adding and removing columns, and now we will see adding and removing rows. For this purpose, we will create two small data sets data1 and data2.

Create a dictionary and convert if into a Pandas data frame.

```python
student1 = {
"Id": ['cs-01', 'mth-02', 'eng-03', 'phy-04', 'mth-05', 'cs-06'],
"Name": ["Fatima", 'Noureen', 'Salman', 'Sana', 'Jack', 'Ali'],
"subject": ['computer science', 'math', 'english', 'physics', 'statistics', 'programming'],
"marks": [45, 90, 60, 37, 49, 50]
}
data1 = pd.DataFrame(student1)
data1
```

Output gives us a data frame with 6 rows and 4 columns.

```python
student2 = {
"Id": ['cs-08', 'mth-0'],
"Name": ["Adam", 'Steve'],
"subject": ['compu' 'science', 'math'],
"marks": [45, 90]
}
data2 = pd.DataFrame(student2)
data2
```

Data frame with 2 rows and 4 columns

Adding rows with Append method
Now lets append data1 and data2 data frames with the append() method.

Append:Pandas append() method appends the rows to the bottom of the first data set.

We will also use ignore_index parameter and set it to True; this resets the index of the appended data set. Otherwise if it is set to False the data frame retains original indexing. Lets see both out puts.

```python
df_resultado = pd.concat([data1, data2], ignore_index=False)
df_resultado
```

```python
df_resultado = pd.concat([data1, data2], ignore_index=True)
df_resultado
```

Adding rows with Concat method
Another method to add rows in pandas is with the concat function.

The concat() method works with columns and rows. It takes an axis parameter which differtnaites the operation. On axis set to rows, the concat method works the same as the append method, but on columns, it is concat by columns, not rows.

Let’s see a practical example

Axis set to 1 the concat operation is for columns

```python
pd.concat([data1, data2], axis=1)
```

NaN values are filler values for places in the data set where there is no data available.

Axis set to 0 the concat operation is for rows

```python
data1 = pd.concat([data1, data2], axis=0, ignore_index=True)
data1
```

We can see the difference in the output when an axis is set to 1 (columns). It concat data2 in data1 by columns and insert NaN values in other rows because we have only two rows in data2.

But on axis 0 (rows) works the same as the append method.

Removing rows
To remove/drop rows, same the drop method is used. Instead of columns parameter, we will use index.

```python
data1.drop(index = [0,7])
```

Row numbers 0 and 7 are deleted. Once again, if we want to change the original data frame, don’t forget to set inplace the parameter equal to True.
