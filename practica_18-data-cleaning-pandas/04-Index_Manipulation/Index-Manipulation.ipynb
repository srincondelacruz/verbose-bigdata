```python
import pandas as pd
```

First, we will see a simple data set created from a dictionary with the Data Frame function.

```python
people = {
"Id": ['cs-01', 'mth-02', 'eng-03', 'phy-04', 'mth-05', 'cs-06'],
"Name": ["Fatima", 'Noureen', 'Salman', 'Sana', 'Jack', 'Ali'],
"subject": ['computer science', 'math', 'english', 'physics', 'statistics', 'programming']
}
data = pd.DataFrame(people)
data
```

Let us review the output of the data (data frame), and we have three columns (Id, Name, subject) and six rows with default index number range(0,1,2,3,….n).



Set custom indexes
Set index method

The set index method is used to set existing columns as the data frame index. Let’s go ahead and make the Id column an index. If we try to access Id like a column (df[‘Id’]), it will show an error. Because once we have set Id as an index, it is no longer a column.

The change of the id column to the index is only temporary, and if we want to make it permanent, we will use inplace=True. This will permanently change our data set.

```python
data.set_index('Id', inplace=True)
data
```

Observe our index is not the Id column and not numbers anymore.

View index method
dataframe.index method is used to view the index of a data frame.It is a quick and easy way to look at the index of a data frame.

```python
data.index
```

Index is Id column as expected, and we can access them with the loc function (discussed here), which is very easy and make more sense than default indexes (0,1,2,3…n).

```python
data.loc['cs-01']
```

Resetting index
Reset index method

The reset_index() method resets the indexes of the data frame and reverts it to the default.

Lets try this method.

```python
data.reset_index(inplace=True)
data
```

Observe Id is now added back as a column.

Set a column as an index
Another method, we can set a custom index with the index_col parameter direct from where we read the data set. Let us see how we can do this with a real-world data set called Stack Overflow Annual Developer Survey. You can download the data set here.

Let’s set the Country column as the index

```python
df = pd.read_csv('../survey_results_public_2020.csv', index_col='Country')
df.head()
```

Observe Country column is now the index.



Index sorting
Sort index method

We can also sort the indexes by axis the sort_index method. By axis mean either we can sort row labels or column labels.

Set axis = 0 to sorts the row labels and axis = 1 for columns. Instead of using axis = 0 or axis = 1 we can also use axis = ‘rows’ or axis = ‘columns’ respectively. If we do not mention an axis it sorts rows index by default.

Ascending order is default. To sort in descending order, we set ascending parameter to False (ascending = False)

As discussed earlier, if we want to change the original data frame, we enable an in place argument by inplace=True.

Let’s sort our row index by descending order.

```python
df.sort_index(ascending=False)
```

Now let’s sort our columns in descending order

```python
df.sort_index(ascending=False, axis='columns')
```

In the above output, you can see that our first column is YearsCodePro and not Respondent anymore.
