```python
import pandas as pd
import numpy as np
```

```python
df_empleados = pd.DataFrame({
 'nombre': ['Ana', 'Juan', 'María', 'Pedro', 'Laura'],
 'edad': [28, 35, 42, 31, 38],
 'salario_mensual': [3000, 4500, 6000, 3800, 5200],
 'años_experiencia': [3, 8, 15, 5, 10],
 'departamento': ['IT', 'Ventas', 'IT', 'Marketing', 'Ventas']
})
```

1. Calcular el salario anual

```python
df_empleados['salario_anual'] = df_empleados['salario_mensual'] * 12
print(df_empleados)
```

2. Calcular el salario por año de experiencia

```python
df_empleados['salario_exp'] = df_empleados['salario_mensual'] / df_empleados['años_experiencia']
print(df_empleados)
```

3. Crear grupos de edad: 'Junior' (< 30), 'Mid' (30-40), 'Senior' (> 40)

```python
bins = [0, 29, 41, 100]
labels = ['Junior', 'Mid', 'Senior']

df_empleados['grupo_edad'] = pd.cut(df_empleados['edad'], bins=bins, labels=labels)
print(df_empleados)
```

4. Crear una variable dummy para el departamento IT

```python
df_empleados['es_IT'] = np.where(df_empleados['departamento'] == 'IT', 1, 0)
print(df_empleados)
```

5. Calcular la mediana de salario por departamento y crear una columna con esta
información

```python
df_empleados['mediana_salario'] = df_empleados.groupby(df_empleados['departamento'])['salario_mensual'].transform('median')
print(df_empleados)
```

Ejercicio 2: Análisis Temporal de Ventas
Crea un dataset de ventas diarias y realiza las siguientes transformaciones:

```python
# Semilla para que nos salgan los mismos números aleatorios
np.random.seed(42)

df_ventas = pd.DataFrame({
    'fecha': pd.date_range('2024-01-01', periods=90, freq='D'),
    'ventas_diarias': np.random.randint(1000, 5000, 90)
})

print(df_ventas.head()) 
```

1. Extraer mes, día de la semana, y número de semana del año

```python
df_ventas['mes'] = df_ventas['fecha'].dt.month
df_ventas['dia'] = df_ventas['fecha'].dt.day_name()
df_ventas['semana'] = df_ventas['fecha'].dt.isocalendar().week
print(df_ventas)
```

2. Crear una variable binaria que indique si es fin de semana

```python
df_ventas['dummy_findesemana'] = np.where((df_ventas['dia'] == 'Saturday') | (df_ventas['dia'] == 'Sunday'), 1, 0)
print(df_ventas)
```

3. Calcular la media móvil de 7 días

```python
df_ventas['media_movil_7d'] = df_ventas['ventas_diarias'].rolling(window=7).mean()
print(df_ventas)
```

4. Crear features de lag de 1, 3 y 7 días

```python
df_ventas['lag_1'] = df_ventas['ventas_diarias'].shift(1)
df_ventas['lag_3'] = df_ventas['ventas_diarias'].shift(3)
df_ventas['lag_7'] = df_ventas['ventas_diarias'].shift(7)
print(df_ventas)
```

5. Calcular el crecimiento porcentual día a día

```python
df_ventas['crecimiento_pct'] = df_ventas['ventas_diarias'].pct_change()
print(df_ventas)
```

6. Identificar si las ventas están por encima o debajo del promedio histórico

```python
df_ventas['media_ventas'] = df_ventas['ventas_diarias'].mean()
df_ventas['rendimiento'] = np.where(df_ventas['ventas_diarias'] > df_ventas['media_ventas'], 'Superior', 'Inferior')
print(df_ventas)
```

Ejercicio 3: Análisis de Clientes E-commerce

```python
df_transacciones = pd.DataFrame({
    'cliente_id': [1, 1, 2, 2, 2, 3, 3, 4, 4, 4, 4],
    'fecha': pd.to_datetime(['2024-01-05', '2024-02-10', '2024-01-15',
                             '2024-02-20', '2024-03-25', '2024-01-08',
                             '2024-03-15', '2024-01-20', '2024-02-05',
                             '2024-02-25', '2024-03-30']),
    'monto': [100, 150, 80, 120, 90, 200, 180, 50, 60, 55, 70],
    'categoria': ['Electrónica', 'Ropa', 'Hogar', 'Electrónica', 'Ropa',
                  'Electrónica', 'Electrónica', 'Hogar', 'Hogar', 'Hogar', 'Hogar']
})

print(df_transacciones)
```

1. Calcular el número total de transacciones por cliente

```python
df_transacciones['transacciones_totales'] = df_transacciones.groupby('cliente_id')['fecha'].transform('count')
print(df_transacciones)
```

2. Calcular el monto total gastado por cliente

```python
df_transacciones['monto_total'] = df_transacciones.groupby('cliente_id')['monto'].transform('sum')
print(df_transacciones)
```

3. Calcular el ticket promedio por cliente

```python
df_transacciones['ticket_cliente'] = df_transacciones.groupby('cliente_id')['monto'].transform('mean')
print(df_transacciones)
```

4. Calcular los días desde la primera compra para cada transacción

```python
df_transacciones['fecha_primera_compra'] = df_transacciones.groupby('cliente_id')['fecha'].transform('min')
df_transacciones['dias_desde_inicio'] = (df_transacciones['fecha'] - df_transacciones['fecha_primera_compra']).dt.days
print(df_transacciones)
```

5. Calcular el tiempo promedio entre compras por cliente

```python
df_transacciones['diferencia_entre_fechas'] = df_transacciones.groupby('cliente_id')['fecha'].diff()
df_transacciones['promedio_entre_fechas'] = df_transacciones.groupby('cliente_id')['diferencia_entre_fechas'].transform('mean')
df_transacciones['diferencia_entre_fechas'] = df_transacciones['diferencia_entre_fechas'].dt.days
df_transacciones['promedio_entre_fechas'] = df_transacciones['promedio_entre_fechas'].dt.days
print(df_transacciones)
```

6. Identificar la categoría más comprada por cada cliente


```python
# Por cada cliente (groupby), x = categoria por cliente, busca la categoría más repetida (.mode()), 
# y si hay un empate, dame la primera que encuentres (.iloc[0])."

df_transacciones['categ_mas_comprada_cliente'] = df_transacciones.groupby('cliente_id')['categoria'].transform(lambda x: x.mode().iloc[0])
print(df_transacciones)
```

7. Crear una variable RFM simplificada:

    R (Recency): días desde la última compra

    F (Frequency): número de compras

    M (Monetary): gasto total

    Aquí solo cacularé la recencia en la siguiente celda haré un rank para concatenar el RFM

```python
fecha_actual = df_transacciones['fecha'].max()
fecha_ultima_compra = df_transacciones.groupby('cliente_id')['fecha'].transform('max')
df_transacciones['recencia_dias'] = (fecha_actual - fecha_ultima_compra).dt.days
print(df_transacciones)
```

```python
df_transacciones['F_rank'] = df_transacciones['transacciones_totales'].rank(method='first', ascending=True)
df_transacciones['M_rank'] = df_transacciones['monto_total'].rank(method='first', ascending=True)
df_transacciones['R_rank'] = df_transacciones['recencia_dias'].rank(method='first', ascending=False)

df_transacciones['RFM_Score'] = (
    df_transacciones['R_rank'].astype(int).astype(str) + 
    df_transacciones['F_rank'].astype(int).astype(str) + 
    df_transacciones['M_rank'].astype(int).astype(str)
)
print(df_transacciones)
display(df_transacciones[['cliente_id', 'R_rank', 'F_rank', 'M_rank', 'RFM_Score']])
```

Ejercicio 4: Feature Engineering con Texto

```python
df_reviews = pd.DataFrame({
    'review': [
        'Excelente producto! Muy recomendado!!!',
        'Horrible, no funciona',
        'Buena calidad y precio justo',
        'NO LO COMPREN, es una ESTAFA',
        'Me encantó, perfecto para lo que necesitaba :)'
    ],
    'rating': [5, 1, 4, 1, 5]
})

print("--- Reseñas Originales ---")
print(df_reviews)
```

1. Calcular la longitud del texto (en caracteres)

```python
df_reviews['longitud'] = df_reviews['review'].str.len()
print(df_reviews)
```

2. Contar el número de palabras

```python
df_reviews['palabras'] = df_reviews['review'].str.split().str.len()
print(df_reviews)
```

3. Calcular la longitud promedio de las palabras

```python

df_reviews['promedio_palabras'] = round(df_reviews['longitud'] / df_reviews['palabras'], 2)
print(df_reviews)
```

4. Contar el número de signos de exclamación

```python
df_reviews['num_exclamacion'] = df_reviews['review'].str.count("!")
print(df_reviews)
```

5. Crear una variable binaria que indique si hay palabras en mayúsculas

```python
df_reviews['dummy_MYUS'] = df_reviews['review'].str.contains(r'[A-Z]', na=False).astype(int)
print(df_reviews)
```

6. Contar el número de emoticones

```python
df_reviews['dummy_emoticonos'] = df_reviews['review'].str.contains(r'[:)]', na=False).astype(int)
print(df_reviews)
```

7. Calcular el sentimiento (positivo/negativo) basado en palabras clave

```python
negativos_pattern = r'(NO|ESTAFA|Horrible)'

df_reviews['dummy_sentimmiento'] = np.where(df_reviews['review'] .str.contains(negativos_pattern, case=False, na=False), 'negativo', 'positivo')
print(df_reviews)
```

8. Crear una variable que indique la intensidad del review (basada en puntuación
y mayúsculas)

```python
df_reviews['intensidad_review(1-10)'] = df_reviews['rating'] * (df_reviews['dummy_MYUS'] + 1)
print(df_reviews)
```

Ejercicio 5: Pipeline Completo de Feature Engineering

```python
# Semilla para reproducibilidad
np.random.seed(42) 

df_prestamos = pd.DataFrame({
    'cliente_id': range(1, 101),
    'edad': np.random.randint(22, 65, 100),
    'ingreso_anual': np.random.randint(20000, 120000, 100),
    'monto_prestamo': np.random.randint(5000, 50000, 100),
    'plazo_meses': np.random.choice([12, 24, 36, 48, 60], 100),
    'historial_crediticio': np.random.choice(['Excelente', 'Bueno', 'Regular', 'Malo'], 100),
    'num_prestamos_anteriores': np.random.randint(0, 10, 100),
    'fecha_solicitud': pd.date_range('2024-01-01', periods=100, freq='D')
})

print("--- DataFrame de Préstamos (Inicial) ---")
print(df_prestamos)
```

1. Ratios financieros relevantes

```python
df_prestamos['ratio_endeudamiento'] = df_prestamos['monto_prestamo'] / df_prestamos['ingreso_anual']
df_prestamos['pago_mensual'] = round(df_prestamos['monto_prestamo'] / df_prestamos['plazo_meses'], 2)
df_prestamos['pago_vs_ingreso'] = df_prestamos['pago_mensual'] / (df_prestamos['ingreso_anual'] / 12)
print(df_prestamos)
display(df_prestamos[['cliente_id', 'ratio_endeudamiento', 'pago_mensual', 'pago_vs_ingreso']])
```

2. Características temporales

```python
df_prestamos['dia_solicitud'] = df_prestamos['fecha_solicitud'].dt.day_name()
df_prestamos['mes_solicitud'] = df_prestamos['fecha_solicitud'].dt.month
df_prestamos['es_fin_de_semana'] = np.where(df_prestamos['fecha_solicitud'].dt.dayofweek >= 5, 1, 0)

print(df_prestamos)
```

3. Codificación de variables categóricas

```python
historial_mapping = {'Malo': 1, 'Regular': 2, 'Bueno': 3, 'Excelente': 4}
df_prestamos['score_historial'] = df_prestamos['historial_crediticio'].map(historial_mapping)
df_dummies = pd.get_dummies(df_prestamos['historial_crediticio'], prefix='historial').astype(int)
df_prestamos = pd.concat([df_prestamos, df_dummies], axis=1)

print(df_prestamos)
```


4. Binning de variables continuas

```python
bins_edad = [20, 35, 50, 70] # Jóvenes, Media, Senior
labels_edad = ['Joven_Riesgo', 'Edad_Media', 'Senior_Estable']
df_prestamos['rango_edad'] = pd.cut(df_prestamos['edad'], bins=bins_edad, labels=labels_edad, right=False)
bins_plazo = [0, 36, 61] # Corto (36 meses o menos) o Largo (más de 36)
labels_plazo = ['Corto_Plazo', 'Largo_Plazo']
df_prestamos['rango_plazo'] = pd.cut(df_prestamos['plazo_meses'], bins=bins_plazo, labels=labels_plazo, right=False)
df_prestamos['rango_ingreso_pct'] = pd.qcut(
    df_prestamos['ingreso_anual'], q=5, labels=['Q1', 'Q2', 'Q3', 'Q4', 'Q5'], duplicates='drop'
)

print(df_prestamos)
```

5. Interacciones entre variables

```python
df_prestamos['ingreso_por_edad'] = df_prestamos['ingreso_anual'] / df_prestamos['edad']
df_prestamos['monto_por_prestamo_previo'] = df_prestamos['monto_prestamo'] / (df_prestamos['num_prestamos_anteriores'] + 1) # +1 para evitar la división por cero

print(df_prestamos)
```


6. Agregaciones relevantes

```python
df_prestamos['riesgo_mora_score'] = df_prestamos['score_historial'] * df_prestamos['num_prestamos_anteriores']
df_prestamos['dias_desde_max_fecha'] = (df_prestamos['fecha_solicitud'].max() - df_prestamos['fecha_solicitud']).dt.days

print(df_prestamos)
```
